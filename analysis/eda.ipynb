```python
numeric_cols = df.select_dtypes(include="number").columns.drop(['repetition','temperature'], errors="ignore")
details = df.groupby(['agent_type','server'])['total_tokens'].agg(['mean','sum']).reset_index()
details['mode'] = details['agent_type'].apply(lambda x: 'code' if (x=='code_mode' or x=='multi_agent_code') else 'mcp')
details['archi'] = details['agent_type'].apply(lambda x: 'multi' if (x=='multi_agent_mcp' or x=='multi_agent_code') else 'single')
details
```

```python
comparison = details.groupby(['server', 'mode','archi'])['mean'].first()
comparison.reset_index()
```

```python
comparison = details.groupby(['server', 'mode','archi'])['mean'].first()
res = comparison.reset_index()
pls = res.groupby(['server','archi'])['mean'].idxmax().reset_index().rename()
res.loc[pls]
```
